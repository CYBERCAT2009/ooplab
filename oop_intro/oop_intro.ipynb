# Знайомство з ООП
Клас `MyName` з виконаними індивідуальними завданнями.

```python
class MyName:
    """Опис класу / Документація
    """
    total_names = 0  # Class Variable

    def __init__(self, name=None) -> None:
        """Ініціалізація класу
        """
        # Якщо ім'я не передали (None) - беремо ім'я з "анонімного" об'єкта
        self.name = name if name is not None else self.anonymous_user().name

        # Перевірка: лише літери
        if not self.name.isalpha():
            raise ValueError("Ім'я може містити лише літери!")

        # Перша літера завжди велика (решту літер не змінюємо)
        self.name = self.name[0].upper() + self.name[1:]

        MyName.total_names += 1  # modify class variable
        self.my_id = self.total_names

    @property
    def whoami(self) -> str:
        """Class property
        return: повертаємо імя
        """
        return f"My name is {self.name}"

    @property
    def my_email(self) -> str:
        """Class property
        return: повертаємо емейл
        """
        return self.create_email()

    @property
    def full_name(self) -> str:
        """Class property
        return: "User #<id>: <name> (<email>)"
        """
        return f"User #{self.my_id}: {self.name} ({self.my_email})"

    def create_email(self, domain="itcollege.lviv.ua") -> str:
        """Instance method
        domain: значення після @ (можна змінити)
        """
        return f"{self.name}@{domain}"

    def count_letters(self) -> int:
        """Instance method: кількість букв в імені
        """
        return len(self.name)

    def save_to_file(self, filename="users.txt") -> None:
        """Instance method: додає рядок із записом у файл
        """
        with open(filename, "a", encoding="utf-8") as file:
            file.write(self.full_name + "\n")

    @classmethod
    def anonymous_user(cls):
        """Classs method
        """
        return cls("Anonymous")

    @staticmethod
    def say_hello(message="Hello to everyone!") -> str:
        """Static method
        """
        return f"You say: {message}"
```

## Створення об'єктів

```python
print("Розпочинаємо створювати обєкти!")

names = ("Bohdan", "Marta", None, "Danylo")  # додано своє ім'я
all_names = {name: MyName(name) for name in names}

for name, me in all_names.items():
    print(f"""{">*<"*20}
This is object: {me}
This is object attribute: {me.name} / {me.my_id}
This is {type(MyName.whoami)}: {me.whoami} / {me.my_email}
This is {type(me.create_email)} call: {me.create_email()}
This is static {type(MyName.say_hello)} with defaults: {me.say_hello()}
This is class variable {type(MyName.total_names)}: from class {MyName.total_names} / from object {me.total_names}
{"<*>"*20}""")

print(f"We are done. We create {me.total_names} names! ??? Why {MyName.total_names}?")
```

## Індивідуальні завдання

```python
print("\n--- Індивідуальні завдання ---")

# 2. Інший текст привітання
print(MyName.say_hello("Привіт, студенте!"))

# 3. Кількість букв в імені
for me in all_names.values():
    print(f"{me.name}: {me.count_letters()} літер(и)")

# 4. Порівняння кількості імен
print(f"Імен у списку names: {len(names)}, лічильник класу: {MyName.total_names}")

# 5. Перша літера завжди велика
print(MyName("danylo").name)

# 6. Інший домен після @
print(all_names["Danylo"].create_email("gmail.com"))

# 7. Перевірка на цифри/символи
for bad_name in ("Dan1lo", "Max!"):
    try:
        MyName(bad_name)
    except ValueError as error:
        print(f"{bad_name}: ValueError - {error}")

# 8. full_name
print(all_names["Danylo"].full_name)

# 9. Запис у файл
for me in all_names.values():
    me.save_to_file()
print("Користувачів збережено у users.txt")
```

створюємо клас, який тепер буде реальним прототипом для створення об'єктів студентів. Клас Student має атрибути name та grade, а також методи get_grade() та set_grade() для отримання та встановлення оцінки студента.
- в класі є особлива функція __init__, яка викликається при створенні об'єкта класу. Вона приймає параметри name та grade, які використовуються для ініціалізації атрибутів об'єкта.
- функції з подвійним підкресленням на початку та в кінці назви є спеціальними методами класу або магічними методами. 

```python
class Student:
    """Це клас, який представляє студента."""
    def __init__(self, name: str, grade: dict, birthday: str = None, age: int = None): # в конструктор передаються аргументи
        self.name = name # атрибут, це поле де можна записати значення та отримати його
        self.grades = grade
        self.birthday = birthday # атрибут, який можна заповнити пізніше
        self.age = age # атрибут, який можна заповнити пізніше

std_1 = Student(name="Любомир Л", grade={"ООП": 90, "Бази даних": 80, "Математика": None, "Українська мова": None})
print(std_1.birthday)  # тут буде None, бо ми не передавали значення для атрибуту birthday
std_2 = Student(name="Василь", grade={"ООП": 30, "Бази даних": 40, "Математика": None, "Українська мова": None})
print(std_2.age)

for student in [std_1, std_2]:
    print(f"Студент {student.name} отримав оцінки:")
    print(f"\tOOП: {student.grades['ООП']}")
    print(f"\tБази даних: {student.grades['Бази даних']}")
    print(f"\tМатематика: {student.grades['Математика']}")
    print(f"\tУкраїнська мова: {student.grades['Українська мова']}")
    print(f"Рейтинг студента {student.name}: {(student.grades['ООП'] + student.grades['Бази даних']) / len(student.grades)}")
```

Атрибути можна перезаписувати, або вони є змінними

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.oop_grades = None
        self.db_grades = None
        self.math_grades = None
        self.ukr_grades = None

s1 = Student("Любомир Л")
s1.oop_grades = 90
s1.db_grades = 80
s2 = Student("Василь")
s2.oop_grades = 30
s2.db_grades = 40

s1.oop_grades = 100 # перезаписуємо значення атрибуту oop_grades для об'єкта s1
for m in [s1, s2]:
    print(f"Студент {m.name} отримав оцінки:")
    print(f"\tOOП: {m.oop_grades}")
    print(f"\tБази даних: {m.db_grades}")
    print(f"\tМатематика: {m.math_grades}")
    print(f"\tУкраїнська мова: {m.ukr_grades}")
    print(f"Рейтинг студента {m.name}: {(m.oop_grades + m.db_grades) / 4}")
```

- для того щоб читати атрибути і не могти їх модифікувати існують властивості (property). Властивість дозволяє створювати методи, які виглядають як атрибути. Властивості дозволяють контролювати доступ до атрибутів та забезпечують інкапсуляцію даних.
- з атрибутами ми працюємо всередині програми а властивості видаємо на зовні, щоб користувачі класу могли отримувати доступ до даних без можливості їх змінювати.

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.oop_grades = None # можна змінювати
        self.db_grades = None

    @property
    def get_oop_grade(self): # властивості можна тільки читати, але не можна змінювати
        return self.oop_grades

    @property
    def get_db_grade(self):
        return self.db_grades

s1 = Student("Любомир Л")
s1.oop_grades = 90
print(f"Оцінка з ООП: {s1.get_oop_grade}")
#s1.get_oop_grade = 100 # тут буде помилка, бо ми не можемо змінювати властивість get_oop_grade
```
