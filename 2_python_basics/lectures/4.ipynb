## Цикли
В Python існують два основні типи циклів: `for` та `while`. Вони дозволяють повторювати блок коду кілька разів, поки виконується певна умова.

### Цикл `for`
Цикл `for` використовується для ітерації по послідовності, ми завжди знаємо кількість ітерацій заздалегідь. Наприклад, ми можемо ітерувати по списку, рядку або діапазону чисел.

### Цикл `while`
Цикл `while` є умовним циклом, який виконує блок коду доти, поки умова є істинною. Він використовується тоді, коли кількість ітерацій невідома заздалегідь.


```python
sl = "Це є елементи"
for i in sl:
    print(f"Елемент: {i}")

ll = [1, 2, 3, 4, 5]
for i in ll:
    print(f"Елемент: {i}")
```

- для генерації послідовності ми використовуємо функцію `range()`, яка створює послідовність чисел від початкового значення до кінцевого (не включно) з кроком, який ми можемо вказати.
- існує безіменнна змінна яка позначається як `_`, її використовують коли нам не потрібне значення змінної в циклі.

```python
print(list(range(3, 10)))  # Виведе: [3, 4, 5, 6, 7, 8, 9]

for i in range(3, 6):
    print(f"Елемент: {i}")

_ = 1
print(f"Значення змінної _: {_}")  # Виведе: Значення змінної _: 1

for _ in range(3):
    print("Цей блок коду виконається 3 рази")

for i in []:
    print("Цей блок коду не виконається жодного разу, бо список порожній")
```

приклад використання циклу `while` - умовного циклу, в якого умова завжди має бути boolean значенням, тобто True або False. Якщо умова є True, то блок коду виконується, якщо False - то ні.
За допомогою While True: ми можемо створити нескінченний цикл, який буде виконуватися доти, поки умова True.

```python
#while True:
#    print("Цей блок коду виконається нескінченно, поки умова True")
```

```python
i = 1
while i <= 5:
    print(f"Елемент: {i}")
    i += 1  # Збільшуємо значення i на 1 після кожної ітерації

print(f"Умова i <= 5: {i <= 5}, тому цикл завершився")
print(f"Умова може бути комплексною: {i},  {i % 2}, {i <= 5 and i % 2 == 0 or i % 3 == 0}, тому цикл завершився")
```

- умова завжди перевіряється перед виконанням блоку коду, тому якщо умова False, то блок коду не виконається жодного разу.
- можна виконати хочаб одну операцію в циклі якщо додати блок else, який виконається після завершення циклу, якщо умова стала False. (остання операція циклу яка виконується при умові False)

```python
while False:
    print("Цей блок коду не виконається жодного разу, бо умова False")
else:
    print("Цей блок коду виконається після завершення циклу, бо умова стала False")
```

Практична задача
симулювати витягування ММДс з пачки та порахувати скільки разів ми витягнули Червоний ММДс

Початкові умови
- визначимо кількість ММДс в пачці (наприклад 20-30)
- крльори в пачці: Червоний, Синій, Зелений, Жовтий, Коричневий
- підрахунок Червоних ММДс, які ми витягнули з пачки

```python
import random

print(f"Кількість ММДс в пачці: {random.randint(20, 30)}")

mmds_colors = ["Червоний", "Синій", "Зелений", "Жовтий", "Коричневий"]

print(f"Виблари ММДс: {random.choice(mmds_colors)}")

if random.choice(mmds_colors) == "Червоний":
    print("УРА! Червоний ММДс")

for mmds in range(random.randint(20, 30)):
    print(f"Вибрали ММДс: {random.choice(mmds_colors)}")
```

```python
n_packs = random.randint(20, 30)
print(f"Кількість ММДс в пачці: {n_packs}")
n_red_mmds = 0 # акамулятор кількості червоних ММДс, які ми витягнули з пачки

for mmds in range(n_packs):
    mmds_color = random.choice(mmds_colors)
    #print(f"Вибрали ММДс: {mmds_color}")
    if mmds_color == "Червоний":
        n_red_mmds += 1
        #print("УРА! Червоний ММДс")

print(f"Ми витягнули {n_red_mmds} Червоних ММДс з пачки, пропорція Червоних ММДс: {n_red_mmds / n_packs * 100:.2f}%")
```

- ми можемо створити таку саму симуляцію з використанням циклу while, де ми будемо витягувати ММДс доти, поки не витягнемо певну кількість Червоних ММДс або поки не закінчаться ММДс в пачці.

```python
n_pack = random.randint(20, 30)
i_n = 1
n_red = 0
print(f"Кількість ММДс в пачці: {n_pack}")

while i_n <= n_pack:
    mmds_color = random.choice(mmds_colors)
    #print(f"Вибрали ММДс {i_n}: {mmds_color}")
    if mmds_color == "Червоний":
        #print("УРА! Червоний ММДс")
        n_red += 1
    i_n += 1

print(f"""
~+Кількість червоних ММДс+~
Червоних: {n_red}, 
Пропорція: {n_red / n_pack * 100:.2f}%
""")
```
