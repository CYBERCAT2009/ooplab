## Розгалуження

```python
import random
condition = random.choice([True, False])

print("Підкиданя монети")
if condition:
    print("Монета випала гербом")
else:
    print("Монета випала решкою")
```

- розгалуженя можуть бути скороченми, коли є лише одна дія, яка виконується у випадку істинності умови. У такому випадку можна не використовувати ключове слово else.

```python
action = random.choice([True, False])
if action:
    print("Дія виконується")
print("Продовження виконання програми")
```

- така скорочена умова найчастіше викорстовується в фнкціях з поєднанням з оператором return. У такому випадку, якщо умова істинна, функція повертає значення і виходить з неї, а якщо умова хибна, то виконання функції продовжується далі.

```python
def classic(num): 
    if num > 0:
        return "Число додатнє"
    else:
        return "Число не додатнє"

def check_number(num):
    if num > 0:
        return "Число додатнє"
    return "Число не додатнє"

print(classic(5), check_number(0))
```

- ми може створити розгалуження використовуючи списки та вбудовані константи

```python
lf = ["Хибність", "Істина"]
print(lf[False], lf[True])

for i in range(5):
    c = random.choice([True, False])
    print(lf[c])
```

```python
l_obj = [min, max]
print(type(l_obj[False]))

print(l_obj[False]([5, 10, 6, 12, 7, 8]), l_obj[True]([5, 10, 6, 12, 7, 8]))
print(min([5, 10, 6, 12, 7, 8]), max([5, 10, 6, 12, 7, 8]))
```

- one-liners - однострічкові вирази, які використовуються для скорочення коду. Вони дозволяють виконувати умовні операції в одному рядку, що робить код більш компактним та зручним для читання.

```python
condition = random.choice([True, False])
a = ""
if condition:
    a = "Умова істинна"
else:
    a = "Умова хибна"

print(a)

a = "Умова істинна" if condition else "Умова хибна"
print(a)
b = 1 if condition else 0
print(b)
```

```python
condition = random.choice([True, False])
n = [1, 5, 7, 3, 9, 2, 4, 6, 8]
nm = min(n) if condition else max(n)
print(f"Випадковий min/max: {nm}")

# спочатку True, потім умова і вкінці False
ns = f"Значення при min: {min(n)}" if condition else f"Значення при max: {max(n)}"
print(ns)
```

- для циклу for ми можемо використовувати однострічкові вирази для створення списків -це називається list comprehension.

```python
# класичний цикл for
n_classic = []
for i in range(5):
    n_classic.append(random.randint(0, i))
    #
    #
print(f"Результат: {n_classic}, останній елемент при виході з циклу: {i}")

# list comprehension
# спочатку маємо тіло циклу, далі власне цикл for
n_list = [random.randint(0, i) for i in range(5)]
```

- а як щодо багатострічкових операцій в тілі циклу - просто робимо функцію

```python
def function_name_for_list_comprehension(i):
    a = random.randint(0, i)
    b = f"Ми маємо число {a} для індексу {i}"
    c = len(b)
    return [a, b, c]


lf = [function_name_for_list_comprehension(i) for i in range(5)]
print(f"Результат: {lf}")
```

- ми можемо робити цикли з розгалуженнями в одній стрічці

```python
rainbow = ["червоний", "помаранчевий", "жовтий", "зелений", "блакитний", "синій", "фіолетовий"]

color_under_7 = [f"Колір: {color}" for color in rainbow if len(color) <= 7]
print(f"Результат: {color_under_7}")
#print(color) # Color не буде існувати поза межами list comprehension
color_above_6 = [f"Колір: {color}" for color in rainbow if len(color) > 6 and color != "фіолетовий"]
print(f"Результат: {color_above_6}")
```

- багаторівневі розгалуження за допомогою if elif else та їх еквівалент у dict

```python
rainbow = ["червоний", "помаранчевий", "жовтий", "зелений", "блакитний", "синій", "фіолетовий"]
color = random.choice(rainbow)

if color == "червоний":
    print("Це теплий колір")
elif color == "помаранчевий":
    print("Це більш тепліший колір")
elif color == "жовтий":
    print("Це колір сонця")
elif color == "зелений":
    print("Це колір природи")
elif color == "блакитний":
    print("Це колір неба")
elif color == "синій":
    print("Це колір води")
else:
    print("Фіолетовий - це холодний колір")

d_colors = {
    "червоний": "Це теплий колір",
    "помаранчевий": "Це більш тепліший колір",
    "жовтий": "Це колір сонця",
    "зелений": "Це колір природи",
    "блакитний": "Це колір неба",
    "синій": "Це колір води",
    "фіолетовий": "Фіолетовий - це холодний колір"
}

print(d_colors[color])
```
