## Виловлювання помилок у коді
- для цього існує конструкція `try...except`, яка дозволяє обробляти винятки, що виникають під час виконання коду.
- є такі програми, називаються linters, які перевіряють код на наявність помилок і попереджень.
- такі помилки є статичними або наперед відомими

```python
#this_string = "Це стрічка' 

if True:
    print("Цей блок коду виконається")
else: # dead code
    print("Цей блок коду ніколи не виконається")
```

```python
import random

list_of_numbers = list(range(-10, 10))
print(list_of_numbers)
for i in range(20):
    li = random.choice(list_of_numbers)
    calculate_element = i / li
    print(f"Результат ділення {i} на {li} = {calculate_element}")
```

- першим способом - це відсіювання граничних варінтів за допомогою конструкції `if...else`, яка дозволяє перевіряти умови і виконувати різні блоки коду залежно від результату перевірки.

```python
for i in range(20):
    li = random.choice(list_of_numbers)
    calculate_element = i / li if li != 0 else None
    print(f"Результат ділення {i} на {li} = {calculate_element}")
```

- якщо ми не знаємо граничних варіантів, то ми повинні використати конструкцію `try...except`, яка дозволяє обробляти винятки, що виникають під час виконання коду.

```python
for i in range(20):
    li = random.choice(list_of_numbers)
    try:
        # Цей блок буде виконуватись при natural flow, якщо не виникне помилка
        calculate_element = i / li
        print(f"Результат ділення {i} на {li} = {calculate_element}")
    except Exception: # це є загальна помилка, яка ловить всі винятки
        # коли ми знаємо про помилку - ми пишемо опрацьовувавч винятку
        print(f"Виникла помилка при значенні {li}")
```

- ми можемо штучно створювати помилки, використовуючи конструкцію `assert`, яка перевіряє умову і викликає виняток, якщо умова не виконується.

```python
should_be_string = "Це стрічка"

assert isinstance(should_be_string, str), "Це не стрічка"
print("Це дійсно стрічка", should_be_string.capitalize())
```

- якщо вже додаємо assert то його не обгортаємо в try except, бо це вже є виняток, який ми хочемо бачити.
- код внизу - так не робимо, бо ми не бачимо помилку, а просто її ловимо і виводимо повідомлення.

```python
try:
    assert isinstance(should_be_string, str), "Це не стрічка"
    print("Це дійсно стрічка", should_be_string.capitalize())
except AssertionError as e:
    print(e)
```

- в основному всі імпорти робляться на початку файлу, але є винятки, коли ми імпортуємо модуль всередині функції, щоб не навантажувати пам'ять, якщо функція не викликається.

```python
#import boto3
# програма буде падати взагалі на почтку
# далі код програми

def boto3_feature():
    try:
        import boto3
        print("Це функція, яка використовує модуль boto3")
    except ImportError as e:
        print(f"WARN! На жаль бібліотека не встановлена: {e}")
        print("Ця фіча виконується без використання boto3")

print("Це основне тіло програми")
condition = random.choice([True, False])
if condition:
    boto3_feature()
else:
    print("Не потрібно використовувати модуль boto3, тому не імпортуємо його")
```
