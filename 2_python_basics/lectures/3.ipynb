## Вбудовані функції
розглядаємо вбудовані функції на прикладі функції print(), яка використовується для виведення даних на екран.
- якщо функція немає дужок - то вона не викликана і ми можемо дізнатися її тип, який буде builtin_function_or_method
- виклик функції здійснюється за допомогою круглих дужок

```python
print( type(print) )
```

```python
print()
print()
print()
```

- те що ми передаємо у функцію називається аргументами, а те що функція повертає називається результатом виконання функції
- для функції print() аргуметом є те що вона виводить на екран
- аргументів може бути декілька, вони передаються через кому
- всі аргументи опрацьовуються і виводяться на екран
- функція print() автоматично додає пробіл між аргументами

```python
print("Студенти!", "Вчити Python", "цікаво")
print("П", "р", "о", "г", "р", "а", "м", "а")
```

- ми можемо попередньо відформатувати аргументи, які передаємо у функцію print() за допомогою оператора +, який об'єднує рядки
- ми формуємо єдиний аргумент для передачі у функцію print() і виводимо його на екран
- повторення - зчеплення стрічок це є конкатенація, яка здійснюється за допомогою оператора +
- при попередньому форматуванні аргументів ми маємо всі дані перевести у str, але якщо це робити через функцію print() то вона автоматично перетворює дані у str і виводить на екран

```python
pf = "Програмування" + " на Python" + " цікаве та корисне"
print(pf)

print("Завдання", "виконано:", 5)

of = "Завдання" + " виконано: " + str(5)
print(of)
```

- початкове форматування стрічки можна виконувати методами join() та format(), які є методами класу str, а також за допомогою f-рядків, які з'явилися у Python 3.6
- join створює одини агрумент для функції print() з елементів списку, які передаються у join()

```python
print("~|~".join(["П", "р", "о", "г", "р", "а", "м", "а"]))
```

```python
v = "Круто!"
i = 5
print("{} найкраща мова програмування, {} {}".format("Python", v, i))
```

- якщо елементи повторюються то ми можемо використати позиційні аргументи у методі format() (будуть появлятись номера позицій у фігурних дужках)

```python
print("{1} найкраща мова програмування, Вчимо {1} на {0}".format(i, "Python"))
```

- найновішим способом форматування стрічок є використання f-рядків (f-strings), які з'явилися у Python 3.12

```python
name = "Python"
fs = f"{name} найкраща мова програмування! Вчимо {name} на {i}"
print(fs)
print(type(fs))
```

Практичний приклад, форматування email за допомогою f-рядків
- ми можемо форматувати email за допомогою f-рядків, які дозволяють вставляти змінні безпосередньо у рядок
- f-рядки можуть бути використані у змінних а далі змінні перевикористані у загальному форматуванні email

```python
import datetime

greeting = "Привіт"
name = "Богдан"
body = "Запрошуємро на курс Python!"
scedule = f"Початок курсу: {datetime.datetime.now()}"
footer = "З повагою, команда Python"

f_email = f"""
{greeting}, {name}!
~~~~~~~~~~~~~~~~~~~
{body}

{scedule}

---
{footer}

{1 + 1} + {2 * 2} = {1 + 1 + 2 * 2}
"""

print(f_email)
```
