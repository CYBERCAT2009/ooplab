# Константи та зарезервовані слова

їх є декілька штук, і найважливішими є:
- True - логічна константа істина
- False - логічна константа хибність
- None - логічна константа відсутність значення

```python
# Ми не можемо перевизначати зарезервовані слова
#True = 1
true = True
print(true, type(true))
tu = "True"
print(tu, type(tu))
```

- True завжди відповідає чомусь в середині, а False завжди відповідає пустоті або незаповненості

```python
if "True":
    print("True відповідає чомусь в середині")

if "":
    print("При пустій стрічці False відповідає пустоті або незаповненості")

if []:  # () , {}
    print("При пустому списку False відповідає пустоті або незаповненості")

if ["Будь-що"]:
    print("При непустому списку True відповідає чомусь в середині")
```

- логічні константи True та False використовуються при порівнянні та логічних операціях
- при логійних операціях результат конвертується до зарезервованих слів True або False

```python
print(1 > 2)
bl = 2 > 1
print(bl, type(bl))

print(1 == True, 5 == True, 0 == False, 1 == False)
if 1:
    print("1 відповідає True")

if not 0: # True або 1
    print("0 відповідає False")
```

- None - це константа, яка відповідає відсутності значення. Вона використовується для позначення порожніх змінних, функцій без повернення значення та інших випадків, коли потрібно вказати на відсутність даних.

```python
ab = print("Друкуємо щось")
print("Результат:", ab)


def my_function():
    #тіло функції - тут виконуються операції
    print("Виконуємо функцію")
    a = 5
    return a

def my_empty_function():
    pass

print("Результат:", my_function())
print("Результат:", my_empty_function())
```

- чи можемо ми використати None в логічних операціях? - так, None вважається False при логічних операціях.

```python
if not None: # False
    print("None відповідає False")

if print("Друкуємо щось"):
    print("Це не виконається, бо print повертає None, а None відповідає False")
```
