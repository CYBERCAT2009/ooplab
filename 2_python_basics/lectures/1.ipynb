Колекції в Python:
- list - список - це колекція впорядкованих елементів, який може містити елементи різних типів, підтримує індексацію і слайсинг, змінюваний (mutable)
- tuple - кортеж, який може містити елементи різних типів, незмінюваний (immutable)

```python
l1 = [1, "2", 3, '4', [1, 2, 3], 5.0]
print(type(l1))
print("Вкладений список: ", type(l1[-2]))
print(l1[3])
l1[3] = "чотири"
print(l1[3])
```

```python
t1 = (1, "2", 3, '4', [1, 2, 3], 5.0)
print(type(t1), t1[3])
#t1[3] = "чотири"
```

Всі властивості колекцій mutable і immutable виходять з того що вони є об'єктами, і у них є методи, які дозволяють змінювати їх стан або не дозволяють цього робити.

```python
print(l1[0])
l1[0] = 100 # це скорочений запис для insert(0, 100)
print(l1[0])
l1.insert(0, 200)
print(l1[0])

print(t1, "Трійка буде на позиції: ", t1.index(3))
```

Через те, що список має методи, які дозволяють змінювати його стан,ми можемо розширювати його, додавати елементи, видаляти елементи, змінювати елементи, а кортеж не має таких методів, тому він незмінюваний.

```python
l1.extend([1, "3", 3, 5])
print(l1)
l1.remove(3) # Видаєляє перший елемент зі значенням 3
print(l1)
```

Наступною колекцією є словники який є іменнованою колекцією, де кожен елемент має ключ і значення, і ми можемо звертатись до елементів за ключем, а не за індексом.

```python
sd = "Колекція" # стрічка
ld = ["К", "о", "л", "е", "к", "ц", "і", "я"] # список

a = 0
b = "0"
dd = {a: "К", b: "о", 2: "л", 3: "е", 4: "к", 5: "ц", 6: "і", 7: "я"} # словник

print(sd[0], ld[0], dd[a], dd[b])
```

У стрічках та списках індекси повинні бути цілими числами, а у словниках ключі можуть бути будь-якого типу, але вони повинні бути унікальними.
При дублікатає - залишається останній ключ, а попередні видаляються.
на позицї ключа або його значення ми можемо поствити будь-який тип даних, навіть колекцію, але ключі повинні бути хешованими (immutable), а значення можуть бути будь-якого типу.
ключ значення ще можна трактувати як поле та 

```python
d1 = {0: "К", 0: "о"} # словник
print(d1[0]) # виведе "о", бо останній ключ 0 замінює попередні
```

Остання колекія - set - множина, яка є колекцією `унікальних` елементів, НЕ впорядкованою і змінюваною. Множини підтримують операції об'єднання, перетину та різниці.

```python
ss = {2, 3, 3, 2, 1, 1, 2, 1, 2, 1} # множина
print(len(ss))  # скільки елементів ми б не задавали, всеодно залишаються лише унікальні елементи
print(ss)  # виводимо множину

# ss[1] - це буде помилка, бо множина не підтримує індексацію, бо вона не впорядкована

print(ss.pop())
ss.add(4)
print(ss)
ss.add(1)
print(ss)
print(ss.pop())
```

### завдання з співбесіди
- як зробити множину з колекції, яка НЕ містить дублікатів
- знайти лише унікальні елементи в колекції
- відсіяти всі дублікати в колекції, залишивши лише унікальні елементи

```python
names = ["Богдан", "Тарас", "Богдан", "Олександр", "Тарас", "Оля", "Ігор", "Оля"]

no_duplicaetes = list(set(names))
print(no_duplicaetes)
```
