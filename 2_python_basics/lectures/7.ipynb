## Узгодження операцій та контекст менеджери
- найкраще це видно при роботі з файлами, де потрібно відкрити файл, виконати операції і закрити його. Контекст менеджери дозволяють автоматично керувати ресурсами, забезпечуючи правильне відкриття та закриття файлів.
- приклад неугодження

```python
fa = open("file.txt", "w")
fa.write("Привіт з першого відритого файла")

fb = open("file.txt", "a")
fb.write("Привіт з другого файла що відкрили ми і хочемо додати текст")
# спеціально закрив після того як ми записали щоб сталась колізія, 
# бо файл вже відкритий і ми його НЕ закрили після першого запису
fb.close()
fa.close()
```

- щоб правильно узгоджувати опереції відкриття та закриття файлів, можна використовувати контекстний менеджер `with`, який автоматично закриває файл після завершення блоку коду.
- структура коду - with (операція яка потребує узгодження) as (ім'я змінної для роботи з ресурсом):

```python
with open("file_with.md", "w") as fw:
    fw.write("Привіт з першого відритого файла")
    fw.write("\n\n\n")
    fw.write("Завершили доступ до файлу\n")
    fw.write("|Число|-|Квадрат|-|Куб|\n")
    fw.write("|---|---|---|---|---|\n")
    for i in range(5):
        fw.write(f"|{i}| - |{i*i}| - |{i*i*i}|\n")
# файл вже закритий після завершення блоку with
#fw.write("Привіт з другого файла що відкрили ми і хочемо додати текст")
```
