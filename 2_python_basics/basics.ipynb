## 1. Типи даних

`str`, `int`, `float`, `bool`, `list`, `tuple`, `dict`, `set`. Функція `type()` показує тип змінної.

```python
a = "змінна з текстом"
a1 = 'a'
b = 1  # числова змінна
b1 = 1.1
c = ["a", 1, 1.25, "Слово", a]  # List
d = {"a": "Слово", "b": 1, a: b}  # Dict
e = ("a", a)  # Tuple
f = {"ss", a + str(b)}  # Set (у прикладі з методички a + b дає помилку: str + int)

for name, value in [("a", a), ("a1", a1), ("b", b), ("b1", b1), ("c", c), ("d", d), ("e", e), ("f", f)]:
    print(f"{name:3} = {value!r:55} -> {type(value).__name__}")
```

```python
# Мій приклад: дані про студента в різних типах
student = "Даниїл"                      # str
lessons = 2                             # int - пар на тиждень
average = 4.6                           # float
is_student = True                       # bool
subjects = ["ООП", "Мережі", "СУБД"]    # list - можна змінювати
schedule = ("Пн", "Вт")                 # tuple - не можна змінювати
marks = {"ООП": 5, "Мережі": 4}         # dict - ключ: значення
tags = {"python", "git", "python"}      # set - тільки унікальні значення

subjects.append("Математика")
marks["СУБД"] = 5

print(student, lessons, average, is_student, schedule)
print("Предмети:", subjects, "| кількість:", len(subjects))
print("Оцінки:", marks, "| оцінка з ООП:", marks["ООП"])
print("Теги (без повторів):", tags)
print("Перетворення типів:", int("42") + 1, str(3.14) + "!", float(lessons))
```

## 3. Вбудовані функції

`abs`, `len`, `sum`, `min`, `max`, `round`, `sorted`, `range`, `type`, `isinstance` працюють без `import`.

```python
print(abs(-12.5), f"є рівним {abs(12.5)}", "і якщо порівняти то: ", abs(-12.5) == abs(12.5))

numbers = [7, 3, 12, 5, 9]
print("len:", len(numbers), "| sum:", sum(numbers), "| min:", min(numbers), "| max:", max(numbers))
print("Середнє:", round(sum(numbers) / len(numbers), 2))
print("sorted:", sorted(numbers), "| у зворотному порядку:", sorted(numbers, reverse=True))
print("isinstance(5, int):", isinstance(5, int), "| type(numbers):", type(numbers))
```

## 4. Цикли

`for` - коли відомо, по чому йдемо; `while` - поки виконується умова. `break` перериває цикл, `continue` переходить до наступного кроку, `else` у циклі виконується, якщо не було `break`.

```python
letters = ["a", "b", "c"]
for i in range(len(letters)):
    print(f"На позиції {i} знаходиться буква {letters[i]}")
else:
    print("Ця конструкція безглузда!")
```

```python
# Мій приклад 1: таблиця множення на 7 (for + range)
for i in range(1, 11):
    print(f"7 x {i:2} = {7 * i:3}")
```

```python
# Мій приклад 2: enumerate і словник
marks = {"ООП": 5, "Мережі": 4, "СУБД": 5}
for n, (subject, mark) in enumerate(marks.items(), start=1):
    print(f"{n}. {subject}: {mark}")
```

```python
# Мій приклад 3: while + break + continue - шукаємо перше число, що ділиться на 7 і 11
n = 0
while True:
    n += 1
    if n % 7 != 0:
        continue          # не ділиться на 7 - пропускаємо
    if n % 11 == 0:
        print("Знайшли:", n)
        break             # виходимо з нескінченного циклу
```

## 5. Розгалуження

`if / elif / else`, тернарний оператор `A if умова else B`, а також `match` (з Python 3.10).

```python
from random import randint
A = randint(0, 1)
print(f"Значить А={A}" if A else "Але може бути що А={}".format(A))
```

```python
# Мій приклад 1: оцінка за балами (if / elif / else)
def grade(points):
    if points >= 90:
        return "A - відмінно"
    elif points >= 75:
        return "B - добре"
    elif points >= 60:
        return "C - задовільно"
    else:
        return "F - не склав"

for p in [95, 80, 61, 40]:
    print(p, "->", grade(p))
```

```python
# Мій приклад 2: парне / непарне в одному рядку
for x in range(1, 6):
    print(x, "парне" if x % 2 == 0 else "непарне")
```

```python
# Мій приклад 3: match - день тижня
for day in [1, 6, 9]:
    match day:
        case 1 | 2 | 3 | 4 | 5:
            print(day, "- робочий день")
        case 6 | 7:
            print(day, "- вихідний")
        case _:
            print(day, "- такого дня немає")
```

## 6. `try` -> `except` -> `finally`

Помилки під час виконання перехоплюємо, щоб програма не зупинялась. `finally` виконується завжди.

```python
A = 0
try:
    print("Що буде якщо", 10/A, "?")
except Exception as e:
    print("Невже це помилка > ", e)
finally:
    print("О це так на тобі!")
```

```python
# Мій приклад: перетворюємо введені рядки в числа, ловимо різні типи помилок
values = ["10", "abc", "0", "5"]
for v in values:
    try:
        result = 100 / int(v)
    except ValueError:
        print(f"'{v}' - це не число")
    except ZeroDivisionError:
        print(f"'{v}' - на нуль ділити не можна")
    else:
        print(f"100 / {v} = {result}")   # виконується, якщо помилки не було
    finally:
        print("  перевірили значення", repr(v))
```

## 7. Контекстний менеджер `with`

`with` сам закриває файл після блоку, навіть якщо всередині сталася помилка.

```python
# Записуємо файл
with open("my_notes.txt", "w", encoding="utf-8") as f:
    f.write("Перший рядок\n")
    f.write("Другий рядок\n")
    for i in range(3):
        f.write(f"Рядок номер {i + 3}\n")

print("Файл закрито?", f.closed)
```

```python
# Читаємо файл з нумерацією рядків
with open("my_notes.txt", "r", encoding="utf-8") as f:
    for n, line in enumerate(f, start=1):
        print(f"{n})> ", line, end="")
```

```python
# Приклад з методички: читаємо README.md цієї роботи (перші 5 рядків)
with open("README.md", "r", encoding="utf-8") as f:
    for _, line in enumerate(f):
        if _ == 5:
            break
        print(f"{_})> ", line, end="")
```

## 8. Лямбда-функції

**Як я розумію лямбда-функції:** це коротка функція без імені, записана в один рядок: `lambda аргументи: вираз`. Вона одразу повертає результат виразу, `return` не потрібен. Зручна, коли функція потрібна один раз - наприклад, як ключ для сортування в `sorted()` або для `map()` / `filter()`. Для складної логіки краще звичайна функція `def`.

```python
def a_b_func(a, b):
    return a, b

this_is_lambda = lambda first, age: f'Цей код написав: {first}, Мені {age:10d} років'
print("Це просто функція:", a_b_func, "\nА це лямбда:", this_is_lambda)
print("Це її виклик:", this_is_lambda('Богдан', 1_00_000))
print(this_is_lambda(*a_b_func("a", 1)))
```

```python
# Мій приклад: звичайна функція і лямбда роблять одне й те саме
def square(x):
    return x ** 2

square_lambda = lambda x: x ** 2
print(square(5), square_lambda(5))

# Лямбда як ключ сортування: сортуємо студентів за балом
students = [("Оля", 82), ("Тарас", 95), ("Ігор", 67)]
print("За балом:", sorted(students, key=lambda s: s[1], reverse=True))

# map і filter
nums = [1, 2, 3, 4, 5, 6]
print("Квадрати:", list(map(lambda x: x ** 2, nums)))
print("Тільки парні:", list(filter(lambda x: x % 2 == 0, nums)))
```
