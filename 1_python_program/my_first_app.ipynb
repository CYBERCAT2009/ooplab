## Відповідь AI/Copilot

**Запит:** *Яку першу програму ти написав би на Python і чи зміг би її пояснити?*

**Відповідь:** гра «Вгадай число».

1. `random.randint(1, 100)` загадує випадкове число;
2. цикл `while True` повторює запит, доки число не вгадано;
3. `input()` зчитує число користувача, `isdigit()` перевіряє, що введено ціле число;
4. `if / elif / else` порівнює й підказує **«Більше!»** або **«Менше!»**;
5. `break` завершує цикл, коли число вгадано.

```python
# Програма, яку запропонував AI/Copilot: гра "Вгадай число".
# У Notebook число вгадує сам комп'ютер методом ділення навпіл.
import random

secret = random.randint(1, 100)
low, high = 1, 100
attempts = 0

print("Я загадав число від 1 до 100.")

while True:
    guess = (low + high) // 2
    attempts += 1
    print(f"Спроба {attempts}: {guess}", end=" -> ")
    if guess < secret:
        print("Більше!")
        low = guess + 1
    elif guess > secret:
        print("Менше!")
        high = guess - 1
    else:
        print(f"Вгадав! Це число {secret}. Кількість спроб: {attempts}")
        break
```
