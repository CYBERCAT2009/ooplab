## Відповідь AI/Copilot

**Запит:** *Яку першу програму ти написав би на Python і чи зміг би її пояснити?*

**Відповідь Copilot:** програма, яка вітається з користувачем.

- `input(...)` показує запитання й чекає, поки користувач введе текст;
- введений текст зберігається у змінній `name`;
- `print(...)` виводить привітання на екран;
- оператор `+` з'єднує рядки.

Наприклад, якщо ввести `Оля`, програма виведе `Привіт, Оля!`

```python
# Програма, яку запропонував GitHub Copilot як "першу програму".
name = input("Як тебе звати? ")
print("Привіт, " + name + "!")
```
