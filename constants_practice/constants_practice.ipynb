Практичне завдання 52: вбудовані константи Python.
Дослідження типів та значень: True, False, None, Ellipsis, __debug__.


```python

# Словник: назва константи -> її значення
constants = {
    "True": True,
    "False": False,
    "None": None,
    "Ellipsis": Ellipsis,
    "__debug__": __debug__,
}

print("Дослідження типів та значень вбудованих констант")
print("-" * 50)

for name, value in constants.items():
    print(f"{name:<10} | значення: {value!r:<10} | тип: {type(value)}")
```
